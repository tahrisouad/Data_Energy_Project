Nettoyage dataset énergie mix

```python
# Renommer la colonne 'Entity' en 'Country'
df_energy = df_energy.rename(columns={'Entity': 'Country'})

# Affichage pour vérifier le changement
df_energy.head(2)
```

```python
# Supprimer les lignes où le 'Code' est manquant (les agrégats)
# On crée une nouvelle variable 'df_energy_clean' pour garder l'originale intacte au cas où
df_energy_clean = df_energy.dropna(subset=['Code']).copy()

# Vérification s'il reste des codes vides
print(f"Nombre de lignes après filtrage : {len(df_energy_clean)}")
print("Nombre de valeurs manquantes dans 'Code' :", df_energy_clean['Code'].isnull().sum())
```

```python
# Remplacer les NaN par 0 dans les colonnes de production d'énergie
energy_cols = ['Oil', 'Gas', 'Coal', 'Hydropower', 'Solar', 'Wind', 'Nuclear', 'Other renewables']

df_energy_clean[energy_cols] = df_energy_clean[energy_cols].fillna(0)

# Vérification que tout est à zéro ou rempli
print("Vérification des NaN après traitement :")
print(df_energy_clean[energy_cols].isnull().sum())
```

```python
# Aperçu final du dataset clean
print("Dimensions finales :", df_energy_clean.shape)
df_energy_clean.head(10)
```

```python
# Reset de l'index pour que ça commence de 0 
df_energy_clean = df_energy_clean.reset_index(drop=True)

# Affichage pour voir le nouvel index
df_energy_clean.head(5)
```

```python
# Sauvegarde du dataframe nettoyé 
# On utilise index=False pour ne pas enregistrer la colonne d'index
df_energy_clean.to_csv('../Data_Clean/energy_mix_clean.csv', index=False)

print("Le fichier energy_mix_clean.csv a été enregistré avec succès dans Data_Clean !")
```

Nettoyage dataset CO2 

```python
#Sélection et Renommage
# Sélection des colonnes utiles et renommage pour la cohérence
cols_to_keep = ['country', 'year', 'iso_code', 'co2', 'co2_per_capita', 'coal_co2', 'gas_co2', 'oil_co2']

df_co2_clean = df_co2[cols_to_keep].copy()

# Renommer pour matcher avec df_energy_clean
df_co2_clean = df_co2_clean.rename(columns={
    'country': 'Country',
    'year': 'Year',
    'iso_code': 'Code'
})

df_co2_clean.head()
```

```python
#Filtrage des Pays
# Garder uniquement les pays (Code ISO non null)
df_co2_clean = df_co2_clean.dropna(subset=['Code']).copy()

print(f"Nombre de lignes après filtrage des pays : {len(df_co2_clean)}")
```

```python
#Imputation par zéro et Index
# Remplacer les NaN par 0 pour les émissions
emissions_cols = ['co2', 'co2_per_capita', 'coal_co2', 'gas_co2', 'oil_co2']
df_co2_clean[emissions_cols] = df_co2_clean[emissions_cols].fillna(0)

# Reset index
df_co2_clean = df_co2_clean.reset_index(drop=True)

# Vérification finale
print("Valeurs manquantes après nettoyage :\n", df_co2_clean.isnull().sum())
df_co2_clean.head()
```

```python
# Sauvegarde du fichier CO2 nettoyé
df_co2_clean.to_csv('../Data_Clean/co2_clean.csv', index=False)

print("Le fichier co2_clean.csv a été enregistré avec succès dans Data_Clean !")
```

Nettoyage de dataset GDP 

```python
# Identifier les colonnes d'années (celles qui sont numériques)
annees = [col for col in df_gdp.columns if col.isdigit()]

# Utiliser pd.melt pour transformer les colonnes d'années en lignes
df_gdp_long = pd.melt(df_gdp, 
                      id_vars=['Country Name', 'Country Code'], 
                      value_vars=annees, 
                      var_name='Year', 
                      value_name='GDP')

# Convertir Year en entier (int) pour le futur merge
df_gdp_long['Year'] = df_gdp_long['Year'].astype(int)

print(f"Dimensions après Melt : {df_gdp_long.shape}")
df_gdp_long.head()
```

```python
# Renommer les colonnes pour la cohérence
df_gdp_clean = df_gdp_long.rename(columns={
    'Country Name': 'Country',
    'Country Code': 'Code'
})

# Supprimer les lignes où le GDP est NaN (facultatif mais recommandé pour alléger)
df_gdp_clean = df_gdp_clean.dropna(subset=['GDP']).copy()

# Reset index
df_gdp_clean = df_gdp_clean.reset_index(drop=True)

print(f"Dimensions finales du GDP : {df_gdp_clean.shape}")
df_gdp_clean.head()
```

```python
# Sauvegarde du fichier gdp nettoyé
df_gdp_clean.to_csv('../Data_Clean/gdp_clean.csv', index=False)

print("Le fichier gdp_clean.csv a été enregistré avec succès dans Data_Clean !")
```

Nettoyage de dataset Pupulation 

Identification des années et Pivotement (Melt)

```python
# Identify year columns (those that are numeric)
years_pop = [col for col in df_pop.columns if col.isdigit()]

# Transform the dataframe from wide to long format
df_pop_long = pd.melt(df_pop, 
                      id_vars=['Country Name', 'Country Code'], 
                      value_vars=years_pop, 
                      var_name='Year', 
                      value_name='Population')

# Preview the result
df_pop_long.head()
```

Renomage et Nettoyage des types

```python
# Rename columns for consistency with other datasets
df_pop_clean = df_pop_long.rename(columns={
    'Country Name': 'Country',
    'Country Code': 'Code'
})

# Convert Year to integer
df_pop_clean['Year'] = df_pop_clean['Year'].astype(int)

# Drop rows with missing population values to keep it clean
df_pop_clean = df_pop_clean.dropna(subset=['Population']).reset_index(drop=True)

# Final check of dimensions
print(f"Final Population Shape: {df_pop_clean.shape}")
df_pop_clean.head()
```

Sauvegarde du fichier

```python
# Save to CSV in the Data_Clean folder
df_pop_clean.to_csv('../Data_Clean/population_clean.csv', index=False)

print("Le fichier population_clean.csv a été enregistré avec succès dans Data_Clean !")
```

Nettoyage de dataset electricity access

```python
# Melting (Transformation)
# Identify year columns
years_acc = [col for col in df_access.columns if col.isdigit()]

# Pivot the table from wide to long
df_acc_long = pd.melt(df_access, 
                      id_vars=['Country Name', 'Country Code'], 
                      value_vars=years_acc, 
                      var_name='Year', 
                      value_name='Electricity_Access')

df_acc_long.head()
```

```python
# Cleaning and Filtering
# Rename columns for consistency
df_acc_clean = df_acc_long.rename(columns={
    'Country Name': 'Country',
    'Country Code': 'Code'
})

# Convert Year to integer
df_acc_clean['Year'] = df_acc_clean['Year'].astype(int)

# VERY IMPORTANT: Drop NaN values (this removes the empty years like 1960-1989)
df_acc_clean = df_acc_clean.dropna(subset=['Electricity_Access']).reset_index(drop=True)

# Final check of the remaining data
print(f"Final Access Data Shape: {df_acc_clean.shape}")
df_acc_clean.head()
```

```python
# Sauvegarde du fichier
df_acc_clean.to_csv('../Data_Clean/electricity_clean.csv', index=False)

print("Le fichier electricity_clean.csv a été enregistré avec succès dans Data_Clean!")
```
